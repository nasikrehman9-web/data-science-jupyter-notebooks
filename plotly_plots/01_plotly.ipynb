```python
# import libraries
import pandas as pd
import plotly.express as px
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
name = ['Aammar', 'Humza', 'Arshad']
height = [5.8, 5.9, 5.7]

# import library
import plotly.express as px

# create a plot (line chart)
fig = px.bar(x=name, y=height, title='Height of Students')
# show the figure
fig.show()


# save the plot in png file



# save as standalone html file
fig.write_html("plot.html")
```

```python
fig.write_image("plot.png")
```

# plots on titanic dataset 

```python
# import dataset 
df = sns.load_dataset('titanic')
# check the dataset
df.head() 
```

```python
# scatter plot 
fig = px.scatter(df, x='age', y='fare', title='plot of age vs fare')
fig.show()
```

```python
# scatter plot 
df.dropna(inplace=True)
fig = px.scatter(df, x='age', y='fare', color='sex', size='fare',title='plot of age vs fare')
fig.show()
fig.write_image("plot_2.png", scale=3) # how to save in 300dpi 
```

```python
import plotly.graph_objects as go

fig = go.Figure(data=go.Scatter(x=[1, 2, 3], y=[4, 5, 6]))
fig.show()
```

```python
# coustomize the plot

fig.update_layout(
    title = 'Line Plot of dummy data ',
    xaxis = {'title': 'Number'},
    yaxis = {'title': 'Value'},
    showlegend = True
)
```

```python
# heatmaps in plotly

data = np.random.randn(20, 20)
heatmap = go.Heatmap(z=data)

fig = go.Figure(data=heatmap)
fig.show()
```

```python
# heatmap on corr of titanic dataset of age and fare using plotly

fig = px.imshow(df[['age', 'fare']].corr())
fig.show()
```
