# sunburst plot to show heirarchy

```python
# create sunburst plot on titanic dataset 
import plotly.express as px
import seaborn as sns 

# load data titanic dataset 
df = sns.load_dataset('titanic')
# Create a sunburst plot
fig = px.sunburst(df, path=['sex', 'class', 'who', 'alive', 'alone'], values='survived', color='sex')

fig.show()
```

```python
tips = sns.load_dataset('tips')
tips_sort = tips.sort_values(by='size', ascending=True)
tips_sort.head()
```

```python
# Create a sunburst plot
fig = px.sunburst(tips_sort, path=['sex', 'smoker', 'day', 'time'], values='size', color='sex')
fig.show()
```
