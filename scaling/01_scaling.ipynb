# 1. Min Max Scaling

```python
import pandas  as pd
from sklearn.preprocessing import MinMaxScaler


# sample dara 
data = {'value': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)
df.head()

```

```python
# scale the data using min max scaler
scaler = MinMaxScaler()
df['value_scaled'] = scaler.fit_transform(df['value'].values.reshape(-1,1))
df.head()
```

# 2. Standard Scalar or Z-score normalization

```python
import pandas  as pd
from sklearn.preprocessing import StandardScaler


# sample dara 
data = {'value': [10, 20, 30, 40, 50]}
df = pd.DataFrame(data)
df.head()
```

```python
# scale the data using min max scaler
scaler = StandardScaler()
df['value_scaled'] = scaler.fit_transform(df['value'].values.reshape(-1,1))
df.head()
```

# 3. Robust Scalar

```python
from sklearn.preprocessing import RobustScaler


# sample dara 
data = {'value': [10, 20, 30, 1000, 50]}
df = pd.DataFrame(data)
# Robsut Scaling
scaler = RobustScaler()
df['scaled_value'] = scaler.fit_transform(df[['value']])
df.head()
```

# 4. logarithmic scaling / normalization

```python
import numpy as np
import pandas as pd

# random data with outliers
data = {'value': [10000, 20000, 30000, 1000000, 50000]}
df = pd.DataFrame(data)

# log tranform
df['value_log'] = np.log(df['value'])
df['value_log2'] = np.log2(df['value'])
df['value_log10'] = np.log10(df['value'])

df.head()
```

```python
import seaborn as sns
sns.boxplot(df, x='value_log10')
```
