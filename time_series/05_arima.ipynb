```python
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA 


# load the sample dataset. we'll use the "AirPassengeers" dataset.
df = pd.read_csv("./resourses/AirPassengers.csv")

# we'll use "pessengers" column to train our ARIMA model.
# for this sample, we'll use ARIMA(2, 1, 2) model.
model = ARIMA(df["#Passengers"], order=(2, 1, 2))

# fit the model.
model_fit = model.fit()

# make predictions.
predictions = model_fit.forecast(steps=10)

print(predictions)

# plot the predictions.
predictions.plot()
```

```python
# EDF test
from statsmodels.tsa.stattools import adfuller
from numpy import log

result = adfuller(df['#Passengers'].dropna())
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])
print('Is data Stationary?', result[1] < 0.05)
```
