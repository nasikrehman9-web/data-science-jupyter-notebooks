# Time Series Data Analysis Project

```python
# import the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
# read the data 
df = pd.read_csv("resourses/AirPassengers.csv")
df.head()
```

# Data Pre-processing 

```python
# change the name of columns 
df.columns = ['Date', 'Number of Passengers']
df.head()
```

```python
sns.set(rc = {'figure.figsize' : (15, 5)})
# make a two sided line plot with plotly
import plotly.express as px
fig = px.line(df, x = 'Date', y = 'Number of Passengers', title = 'Number of Passengers over Time')
fig.update_xaxes(tickangle = 45)
fig.show()
```

```python
# decompose the time series
from statsmodels.tsa.seasonal import seasonal_decompose
result = seasonal_decompose(df['Number of Passengers'], model = 'multiplicative', period = 12)
result.plot()
# plot in plotly
import plotly.express as px
fig = px.line(df, x = 'Date', y = 'Number of Passengers', title = 'Number of Passengers over Time')
fig.update_xaxes(tickangle = 45)
fig.show()
```
