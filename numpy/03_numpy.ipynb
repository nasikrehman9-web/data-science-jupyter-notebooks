# Numpy indexing and slicing

```python
import numpy as np 
```

```python
# create an array 
a = np.array([1,2,3,4,5,6,7])
a
```

```python
a[1:5]
```

```python
b = np.arange(0, 12, 2)
b
```

```python
b[2]
```

```python
c = b.reshape(3, 2)
c
```

```python
c[0]
```

```python
c[(c < 5) & (c>7)]
c
```
