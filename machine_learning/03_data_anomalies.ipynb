```python
import pandas as pd
```

```python
data = {
    'date': ['2021-12-01', '01-12-2022', '2022/12/01', '12-01-2021'],
    'country': ['USA', 'U.S.A.', 'America', 'United States'],
    'name': ['Aammar', 'Amaar', 'Hamza', 'Hazma'],
    'sales_2020': [100, 200, None, 200],
    'sales_2021': [None, 150, 300, 150]
}
# make pandas dataframe
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
# strandardizing the date format
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['date'] = df['date'].dt.strftime('%Y-%m-%d')
df.head()
```

# impute fill the date and try another method to complete the data format inconsistencies 

```python
# impute fill the date
df['date'] = df['date'].ffill()
df.head()
```

```python
# harmonizing names of countries 
df['country'] = df['country'].replace({
    'U.S.A.': 'United States',
    'America': 'United States',
    'USA': 'United States'
})
df.head()
```

```python
# Correct the typographical mistakes in name '
df['name'] = df['name'].replace({
    'Amaar': 'Aammar',
    'Hazma': 'Hamza'
})
df.head()
```

```python
# remove the duplicates
df = df.drop_duplicates(subset=['name'])
df.head()
```

```python
df.head()
```

```python
# resolving contradictory data
# For demonstration, let's assume sales_2021 should always be heigher than sales_2020
# we'll remove rows where this condition is not met
df = df.drop(df[df['sales_2021'] <= df['sales_2020']].index)
df.head()
```
