# Support vector machines
support vector machines (SVM) is a type of supervised learning algorithm that uses a set of hyperplanes to separate different classes in a high-dimensional space. It is a linear classifier, meaning that the decision boundary is a straight line in the feature space. 

types of SVM kernels:

- linear kernel
- polynomial kernel
- radial basis function (RBF) kernel

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ml lararies
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder
```

```python
# load the data 
df = sns.load_dataset('iris')
df.head()
```

```python
# lets make X and y in our data 
X = df.drop('species', axis=1)
y = df['species'] 
```

```python
# train test split the data  
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20,random_state=42)
```

```python
# call the model 
model = SVC(kernel='rbf')
# train the model 
model.fit(X_train, y_train)
# predict the mdoel 
y_pred = model.predict(X_test) 
```

```python
# evaluate the model
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
# draw the confusion matrix using heatmap 
plt.figure(figsize=(12, 8))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True)
# label the plot 
plt.xlabel('predicted')
plt.ylabel('Truth')
plt.show()
```
